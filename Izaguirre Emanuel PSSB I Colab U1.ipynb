- *¿Cuáles son las fases de la etapa de Análisis (Front-End) de un compilador?*  
**RESPUESTA:**

Las fases principales que componen esta etapa son
:
Análisis léxico (o lexicográfico / exploración): Es la primera fase del compilador
. Se encarga de leer la cadena de caracteres del programa fuente de izquierda a derecha y agruparlos en unidades lógicamente coherentes denominadas tokens o componentes léxicos (como palabras reservadas, identificadores, constantes y operadores)
. Durante este proceso también se suelen descartar comentarios y espacios en blanco
.
Análisis sintáctico (o parsing): Toma la secuencia de tokens devuelta por el analizador léxico y comprueba que esté bien formada según las reglas gramaticales del lenguaje de programación
. Construye como resultado una representación jerárquica del código, conocida como árbol sintáctico
.
Análisis semántico: Revisa el árbol sintáctico y consulta la tabla de símbolos para verificar que el programa respete las reglas de significado del lenguaje
. Su función principal incluye la comprobación (verificación) de tipos, el control del rango de valores y la validación de que las variables y operandos sean compatibles en cada operación
.
En el diseño estructural de compiladores, el concepto ampliado de Front-End abarca todas las fases que dependen únicamente del lenguaje fuente, integrando formalmente la generación de código intermedio como la fase de salida que conecta con la etapa de Síntesis (Back-End)

- *¿Qué diferencia a un Lexema de un Token?*  
**RESPUESTA:**

La diferencia fundamental entre un lexema y un token radica en su nivel de abstracción y en el papel que desempeñan durante el análisis léxico de un compilador:
Lexema: Es la secuencia concreta de caracteres en el código fuente que coincide con la regla o patrón definido para una categoría del lenguaje
. Representa la cadena de texto real que lee el analizador léxico (por ejemplo, cadenas como posicion, 3.1416, + o while)
.
Token (o componente léxico): Es la unidad abstracta o categoría lógica asociada a un patrón que el analizador léxico entrega al analizador sintáctico
. Por lo general, se compone de un nombre o tipo de token (símbolo abstracto) y, opcionalmente, un valor de atributo (como un apuntador a la tabla de símbolos o el valor de una constante)
.
Principales diferencias:
Concreto vs. Abstracto: El lexema es la cadena de caracteres del programa fuente, mientras que el token es el símbolo abstracto que utiliza la gramática del compilador para realizar el análisis sintáctico
.
Entrada vs. Salida: El lexema es la cadena de caracteres identificada en el texto fuente de entrada, mientras que el token (representado habitualmente como <nombre_token, valor_atributo>) es la estructura de salida que se envía a la siguiente fase del compilador
.
Relación Muchos a Uno: Para categorías genéricas como id (identificador) o numero, un solo token abarca múltiples lexemas distintos (por ejemplo, los lexemas x, suma y total pertenecen todos al mismo token id)
. Sin embargo, para palabras reservadas (como if o while) existe una correspondencia directa entre el lexema y su token
.

- *¿Qué función cumple la Tabla de Símbolos durante la compilación?*  
**RESPUESTA:**

La Tabla de Símbolos (también llamada tabla de nombres o diccionario de datos) es una estructura de datos central y de alto rendimiento que actúa como repositorio durante todo el proceso de compilación
. Su función principal es registrar los identificadores creados por el usuario (como variables, funciones, constantes, tipos y procedimientos) y recolectar sus atributos para ponerlos a disposición de todas las fases del compilador
.
Funciones principales por fase de compilación:
Análisis léxico:
Registro e identificación: Cuando el analizador léxico reconoce un identificador (lexema), consulta la tabla de símbolos para verificar si ya existe. Si no está, crea una entrada nueva y devuelve como atributo del token un puntero o índice que apunta a dicha entrada
.
Manejo de palabras reservadas: En muchos diseños, se utiliza para almacenar previamente las palabras clave del lenguaje, lo que permite diferenciarlas fácilmente de los identificadores de usuario
.
Análisis sintáctico y gestión de ámbitos (Scope):
Vincula las apariciones de los identificadores en la estructura del código con sus declaraciones correspondientes
.
Gestiona la regla del bloque anidado más cercano, apilando o encadenando tablas de símbolos para diferenciar variables globales y locales en lenguajes con estructura de bloques
.
Análisis semántico:
Comprobación de tipos: Permite al analizador semántico verificar que los operandos utilizados en expresiones y asignaciones sean compatibles
.
Validación de contexto: Verifica que no existan variables sin declarar, redeclaraciones no permitidas en el mismo ámbito o usos incompatibles con la categoría del símbolo
.
Generación y optimización de código:
Aporta la información técnica necesaria para la etapa de síntesis, como la dirección de memoria en tiempo de ejecución, desplazamientos (offsets), tamaño de los datos y registros asignados
.
Información típica que almacena:
Nombre/Lexema: La cadena de caracteres del identificador
.
Categoría: Si es variable, constante, función, procedimiento, parámetro o tipo definido
.
Tipo de dato: Entero, real, booleano, puntero, arreglo, etc.
.
Ubicación en memoria: Dirección relativa o desplazamiento para el tiempo de ejecución
.
Detalles de subrutinas: Número y tipo de argumentos, método de paso de parámetros y tipo de valor devuelto
.
Nota: La tabla de símbolos permanece activa en memoria durante la compilación y se descarta al finalizar la generación del programa objeto (salvo si se incluyen opciones de depuración)
. En el caso de los intérpretes, se conserva en memoria durante toda la ejecución

## El "Monstruo Completo" (Enlace Estático)  
Por defecto, GCC usa ***enlace dinámico*** usando una estructura de datos tabular, la tabla **PLT** (*Procedure Linkage Table*) ya que el código usa `printf`, el programa necesita conectarse con la biblioteca estándar de C (`libc.so`). El enlazador no copia el código de `printf` dentro del archivo para no duplicar espacio en el disco; en su lugar, crea un "trampolín" o acceso directo en una sección llamada `<printf@plt>`.

Cada vez que se invoca a `printf`, realmente salta a este bloque plt, el cual averigua en qué parte de la memoria RAM del sistema operativo se encuentra la verdadera función `printf` y redirige el control allí en tiempo de ejecución.

Para ver cómo el enlazador inyecta literalmente miles y miles de líneas copiando físicamente el código de `printf` y todas sus dependencias dentro del ejecutable, se debe compilar de forma estática:

```bash
# Compilación estática
!gcc -static hola_compilador.c -o hola_estatico

# Conteo de líneas de código ensamblador
!objdump -d hola_estatico | wc -l

````python
import token
import tokenize
from io import BytesIO

# Código fuente de prueba en formato cadena
codigo_fuente = "suma = a + 10"

# Tokenización del código fuente
tokens = tokenize.tokenize(BytesIO(codigo_fuente.encode("utf-8")).readline)

print(
    f"{'LINEA/COL':<12} | {'TIPO DE TOKEN':<20} | {'VALOR (LEXEMA)':<15}"
)
print("-" * 55)
for tok in tokens:
    if tok.type in (
        tokenize.ENCODING,
        tokenize.ENDMARKER,
        tokenize.NL,
        tokenize.NEWLINE,
    ):
        continue
    nombre_token = token.tok_name[tok.type]
    posicion = f"{tok.start[0]}:{tok.start[1]}"
    print(f"{posicion:<12} | {nombre_token:<20} | {tok.string:<15}")
````

## 5. Sección Práctica 3: Inspección de la Tabla de Símbolos y Bytecode
En esta sección simularemos la interacción con la Tabla de Símbolos y observaremos la generación de código intermedio.

### [CÓDIGO 3.1] Simulación de una Tabla de Símbolos básica en Python
```python
class TablaDeSimbolos:

    def __init__(self):
        self.simbolos = {}

    def insertar(self, nombre, tipo, valor=None, ambito="global"):
        if nombre in self.simbolos:
            print(f"[ERROR LÉXICO/SINTÁCTICO] Identificador '{nombre}' ya declarado.")
        else:
            self.simbolos[nombre] = {
                "tipo": tipo,
                "valor": valor,
                "ambito": ambito,
            }
            print(f"[TABLA DE SÍMBOLOS] Insertado: {nombre} ({tipo})")

    def buscar(self, nombre):
        return self.simbolos.get(nombre, None)

    def mostrar(self):
        print("\n=== CONTENIDO DE LA TABLA DE SÍMBOLOS ===")
        print(f"{'NOMBRE':<12} | {'TIPO':<10} | {'ÁMBITO':<10} | {'VALOR':<10}")
        print("-" * 50)
        for nombre, datos in self.simbolos.items():
            print(
                f"{nombre:<12} | {datos['tipo']:<10} | {datos['ambito']:<10} | {str(datos['valor']):<10}"
            )


# Prueba de la Tabla de Símbolos
ts = TablaDeSimbolos()
ts.insertar("a", "ENTERO", 5)
ts.insertar("b", "ENTERO", 10)
ts.insertar("suma", "ENTERO", 15)
ts.mostrar()
```


````python
class TablaDeSimbolos:

    def __init__(self):
        self.simbolos = {}

    def insertar(self, nombre, tipo, valor=None, ambito="global"):
        if nombre in self.simbolos:
            print(f"[ERROR LÉXICO/SINTÁCTICO] Identificador '{nombre}' ya declarado.")
        else:
            self.simbolos[nombre] = {
                "tipo": tipo,
                "valor": valor,
                "ambito": ambito,
            }
            print(f"[TABLA DE SÍMBOLOS] Insertado: {nombre} ({tipo})")

    def buscar(self, nombre):
        return self.simbolos.get(nombre, None)

    def mostrar(self):
        print("\n=== CONTENIDO DE LA TABLA DE SÍMBOLOS ===")
        print(f"{'NOMBRE':<12} | {'TIPO':<10} | {'ÁMBITO':<10} | {'VALOR':<10}")
        print("-" * 50)
        for nombre, datos in self.simbolos.items():
            print(
                f"{nombre:<12} | {datos['tipo']:<10} | {datos['ambito']:<10} | {str(datos['valor']):<10}"
            )


# Prueba de la Tabla de Símbolos
ts = TablaDeSimbolos()
ts.insertar("a", "ENTERO", 5)
ts.insertar("b", "ENTERO", 10)
ts.insertar("suma", "ENTERO", 15)
ts.mostrar()
````

2. Implementa una función de prueba preliminar (o utiliza la librería `tokenize` de Python como simulador) que tome la cadena de tu lenguaje y la desglose en una lista de componentes léxicos.

````python
import re

def tokenizar_robolang(ruta_archivo):
    # Definición de patrones de tokens usando expresiones regulares
    patrones = [
        ('COMMENT', r'#.*'),
        ('NUMBER', r'\b\d+\b'),
        ('ASSIGN', r'='),
        ('COMPARE', r'=='),
        ('KEYWORD', r'\b(INIT_ROBOT|FORWARD|ROTATE|IF|THEN|END|STOP)\b'),
        ('IDENTIFIER', r'\b[a-zA-Z_][a-zA-Z0-9_]*\b'),
        ('NEWLINE', r'\n'),
        ('SKIP', r'[ \t]+'),
        ('MISMATCH', r'.'),
    ]

    # Combinar patrones en una sola expresión regular
    regex = '|'.join(f'(?P<{name}>{pattern})' for name, pattern in patrones)

    # Leer el archivo de código fuente
    with open(ruta_archivo, 'r') as f:
        codigo = f.read()

    print(f"{'TIPO DE TOKEN':<15} | {'VALOR (LEXEMA)':<20}")
    print("-" * 40)

    for match in re.finditer(regex, codigo):
        tipo = match.lastgroup
        valor = match.group()
        if tipo == 'SKIP' or tipo == 'COMMENT' or tipo == 'NEWLINE':
            continue
        elif tipo == 'MISMATCH':
            print(f"[ERROR LÉXICO] Caracter no válido: {valor}")
        else:
            print(f"{tipo:<15} | {valor:<20}")

tokenizar_robolang('programa_original.robo')
````

3. Muestra una estructura en código que sirva como **Tabla de Símbolos inicial** donde se registren las variables declaradas en tu lenguaje.

````python
class TablaSimbolosRoboLang:
    def __init__(self):
        # Diccionario para almacenar las variables y sus atributos
        self.simbolos = {}

    def declarar(self, nombre, tipo, valor_inicial=None):
        if nombre in self.simbolos:
            print(f"[ADVERTENCIA] La variable '{nombre}' ya está registrada. Actualizando valor...")
            self.simbolos[nombre]['valor'] = valor_inicial
        else:
            self.simbolos[nombre] = {
                'tipo': tipo,
                'valor': valor_inicial,
                'ambito': 'global'
            }
            print(f"[TABLA DE SÍMBOLOS] Variable '{nombre}' ({tipo}) registrada con éxito.")

    def mostrar_tabla(self):
        print("\n=== TABLA DE SÍMBOLOS INICIAL (RoboLang) ===")
        print(f"{'VARIABLE':<12} | {'TIPO':<10} | {'ÁMBITO':<10} | {'VALOR':<10}")
        print("-" * 50)
        for var, info in self.simbolos.items():
            print(f"{var:<12} | {info['tipo']:<10} | {info['ambito']:<10} | {str(info['valor']):<10}")

# Simulación de registro del análisis de 'programa_original.robo'
tabla_robo = TablaSimbolosRoboLang()

# Registramos las variables encontradas en nuestro código
tabla_robo.declarar("SPEED", "ENTERO", 50)
tabla_robo.declarar("SENSOR", "ENTERO", 1)

# Visualizamos el estado de la tabla
tabla_robo.mostrar_tabla()
````

1. **Pregunta 1 (Conflictos Léxicos):** Al revisar las palabras reservadas y los identificadores de tu lenguaje, ¿existe alguna regla léxica que pudiera causar ambigüedad (por ejemplo, que una palabra reservada coincida con el patrón de un identificador de usuario)? ¿Cómo la resolverá tu analizador?

**RESPUESTA:**

En **RoboLang**, al igual que en la mayoría de los lenguajes de programación, existe una posible ambigüedad léxica: las palabras reservadas (como `FORWARD`, `ROTATE`, `IF`, `THEN`) cumplen con el mismo patrón sintáctico que los identificadores de usuario (definidos como secuencias alfanuméricas que comienzan con letra).

**Resolución del conflicto:**
Nuestro analizador léxico resuelve esta ambigüedad aplicando dos reglas estándar de prioridad:
1. **Prioridad por Orden de Declaración (Precedencia de Reglas):** En el motor de expresiones regulares de nuestro tokenizador, definimos la regla de `KEYWORD` antes de la regla de `IDENTIFIER`. De este modo, si la cadena analizada coincide con ambos patrones, el lexer siempre la reconocerá y clasificará prioritariamente como una palabra clave.
2. **Coincidencia Exacta de Palabra Completa (Word Boundaries):** Usamos los límites de palabra `\b` en las expresiones regulares para evitar subcoincidencias incorrectas (por ejemplo, asegurar que la variable `FORWARD_SPEED` no se confunda o divida erróneamente con la palabra reservada `FORWARD`).
